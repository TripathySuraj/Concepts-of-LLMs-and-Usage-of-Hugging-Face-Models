##**I. Using Google Studio API key and Genai client**
**Getting Gemini Api Key**

```python
from google.colab import userdata

# Fetch API key from Colab Secrets
gemini_key = userdata.get('GEMINI_API_KEY')

print(type(gemini_key))
```

**Initializing client and choosing model**

```python
from google import genai

# Initialize Gemini
client = genai.Client(api_key=gemini_key)

# Select Gemini Flash-Lite
MODEL = "gemini-3.5-flash-lite"

print("Gemini connected successfully!")
```

**Creating a function to pass prompt from the llm model**

```python
def ask_gemini(prompt):
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )
    return response.text
```


##**III. Content Writer using Groq client**
**1. Installing groq and Initializing Groq Client**

```python
# installing groq
!pip install -q groq

from groq import Groq

client = Groq(
    api_key=userdata.get("LLMA_KEY")
)
```

**2. Instructions and Input from user**

```python
# Define AI Assistant Instructions
instructions = """
You are a professional AI Content Writer and Article Writing Assistant.

Your responsibility is to convert the user's question or topic
into a well-structured, informative and engaging article.

Follow these 5 instructions strictly:

1. CONTENT QUALITY:
   Generate meaningful, accurate, informative and relevant content.
   Avoid unnecessary information and repetitive sentences.

2. STRUCTURE AND FORMATTING:
   Write the content like a professional article.
   Include a suitable title, introduction, multiple headings,
   subheadings and a conclusion.

3. CLARITY AND READABILITY:
   Use simple, professional and easy-to-understand English.
   Explain complex concepts in a beginner-friendly manner.

4. CONTENT DEPTH:
   Cover the important aspects of the given topic.
   Include practical examples, suggestions or use cases wherever
   relevant. Do not provide superficial explanations.

5. PROFESSIONAL PRESENTATION:
   Maintain a professional and engaging writing style.
   Use bullet points, numbered lists and paragraphs wherever
   appropriate. Avoid unsupported claims.

ARTICLE FORMATTING RULES:

- Start with a meaningful article title using a Markdown H1 heading.
- Write an engaging introduction.
- Divide the main content into logically arranged sections.
- Use H2 and H3 headings for proper content hierarchy.
- Separate major sections with Markdown horizontal boundaries (---).
- End with a meaningful conclusion.
- Do not include unnecessary introductory comments about being an AI.
- Return the article in clean Markdown format.
"""

print("===== AI CONTENT WRITER ASSISTANT =====")

# Take User Input
question = input("\nEnter your question or article topic: ")
```

**3. Generating and extracting content**

```python
# Generating Content

print("\nGenerating your article... Please wait.")

completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "system",
            "content": instructions
        },
        {
            "role": "user",
            "content": f"Write a detailed article on: {question}"
        }
    ]
)

# Extract generated content
article = completion.choices[0].message.content
```

**4. Saving content in .md file**

```python
# Save as Markdown File
import re

filename = re.sub(r'[^a-zA-Z0-9]+', '_', question.strip())
filename = filename.strip('_')[:60].lower()

if not filename:
    filename = "generated_article"

filename += ".md"

with open(filename, "w", encoding="utf-8") as file:
    file.write(article)

print("\nArticle successfully saved!")
print("Filename:", filename)
```

##**IV. Chain of Thought Prompting**
* ***Generate an initial response using the first prompt, then pass that response back to the AI along with additional instructions to review, refine, and improve it to produce a final, more accurate and well-structured response.***

**1. Planning structure and creating content**

```python
# Planning the structure
topic = "What is LLM Knowledge Cutoff?"

planning_prompt = f'''You're an exprienced content writer, who is specialized in writing content on tech specially AI/Data science.
You explain topics that are usually difficult to explain with real life examples. Your task is to plan the structure of the article, you don't need to write it.
Make a detailed structure plan of the article on the given topic.

Topic = {topic}'''

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": planning_prompt
        },
    ]
)

planning_structure = response.choices[0].message.content


# Creating actual content on basis of given planning structure
content_writer_prompt = f'''You're an exprienced content writer. You will write content on the given plan. Make sure to write content which is wthinically correct.
Your palnning structure for this is : {planning_structure}

Write detailed article as per this plan in markdown format'''

response = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
        {
            "role": "user",
            "content": content_writer_prompt
        },
    ]
)

final_content = response.choices[0].message.content
```

**2. Saving content in .md file**

```python
# Save as Markdown File
import re

filename = re.sub(r'[^a-zA-Z0-9]+', '_', topic.strip())
filename = filename.strip('_')[:60].lower()

if not filename:
    filename = "generated_article"

filename += ".md"

with open(filename, "w", encoding="utf-8") as file:
    file.write(final_content)

print("\nArticle successfully saved!")
print("Filename:", filename)
```

#**V. Conversational Memory in LLMs**
* **Saving history of communication and using them in further communications for context like in case of Chatgpt**

```python
history = []

print("-------- HI There --------")

while True:
    query = input("\nEnter your query: ")

    # Convert previous history into string context
    context = ""

    for communication in history:
        context += "\nUser: " + communication["User"]
        context += "\nAI: " + communication["AI"]

    # Add current query
    context += "\nUser: " + query

    # Generate LLM response
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {
                "role": "user",
                "content": context
            }
        ]
    )

    llm_response = response.choices[0].message.content

    # Store conversation in history
    communication = {
        "User": query,
        "AI": llm_response
    }
    history.append(communication)

    print("\n-------- Response --------")
    print(llm_response)

    # Ask whether the user wants to continue
    choice = input("\nDo you want to communicate further? (yes/no): ")

    if choice.lower() not in ["yes", "y"]:
        print("\n-------- Thank You! --------")
        break
```
